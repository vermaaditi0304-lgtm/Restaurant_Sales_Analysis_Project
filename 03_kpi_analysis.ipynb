```python
import pandas as pd
df = pd.read_csv('../data/cleaned/restaurant_featured.csv', parse_dates=['order_date'])
```

```python
total_revenue = df['total_amt'].sum()
print(f"Total Revenue: ₹{total_revenue:,}")
```

```python
total_orders = df['order_id'].nunique()
print(f"Total Orders: {total_orders:,}")
```

```python
aov = df['total_amt'].mean()
print(f"Average Order Value (AOV): ₹{aov:.2f}")
```

```python
rev_per_customer = df.groupby('customer_id')['total_amt'].sum()
print(f"Average Revenue per Customer: ₹{rev_per_customer.mean():,.2f}")
print(f"Top Spender: {rev_per_customer.idxmax()} — ₹{rev_per_customer.max():,}")
```

```python
monthly = df.groupby(df['order_date'].dt.to_period('M'))['total_amt'].sum()
monthly_growth = monthly.pct_change() * 100
print(monthly_growth.tail(6).round(2))
```

```python
print(df['category'].value_counts())
```

```python
print(df.groupby('age_group')['total_amt'].sum().sort_values(ascending=False))
```

```python
print(df['payment_method'].value_counts(normalize=True).mul(100).round(1))
```

```python
time_rev = df.groupby('time_of_sale')['total_amt'].sum().sort_values(ascending=False)
print("Revenue by time slot:")
print(time_rev)
```

```python
city_rev = df.groupby('city')['total_amt'].sum().sort_values(ascending=False)
city_pct = (city_rev / city_rev.sum() * 100).round(1)
print(pd.DataFrame({'Revenue': city_rev, 'Share %': city_pct}))
```
